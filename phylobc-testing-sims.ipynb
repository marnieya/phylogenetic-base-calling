```python
# getting the paths to specific pileup files from the simulations (variable error)
verr_subfolders = subprocess.run('ls ' + verr_sims_folder_path, shell=True, stdout=subprocess.PIPE).stdout.decode('utf-8')
verr_subfolders = verr_subfolders.strip().split('\n')
print(verr_subfolders)
```

```python
# getting the paths to specific pileup files from the simulations (constant error)
cerr_subfolders = subprocess.run('ls ' + cerr_sims_folder_path, shell=True, stdout=subprocess.PIPE).stdout.decode('utf-8')
cerr_subfolders = cerr_subfolders.strip().split('\n')
print(cerr_subfolders)
```

```python
# running accuracy in the constant error simulations (at 0.5% -- i.e. ~150 bp will be wrong)
# version of the tree where we prune out the target genome OUXXXX.X
accuracy_cmd = open('get_accuracies_cerr.sh','w')
accuracy_cmd.writelines('#!/bin/bash\n')

def get_cmd(subfolder):
    pileup_file = subfolder[12:] + '_sim_piledup.txt'
    pileup_results_path = cerr_sims_folder_path + subfolder + "/"
    return ' '.join(["./" + phylobc_script, pileup_file, pileup_results_path, 't'])

run_script_cmds = [get_cmd(sf) for sf in cerr_subfolders]
accuracy_cmd.writelines(' & \n'.join(run_script_cmds))
accuracy_cmd.close()
```

```python
subprocess.run('chmod +x get_accuracies_cerr.sh', shell=True)
subprocess.run('./get_accuracies_cerr.sh >> cerr_accuracies.txt', shell=True)
```
